# INFRA-RISK Tunisia — demo exploration

This notebook inspects the reproducible demo velocity raster and ranked assets. Demo values are synthetic, not satellite observations.

```python
from pathlib import Path
import rasterio
import matplotlib.pyplot as plt
import pandas as pd

raster_path = Path('../data/processed/demo_velocity.tif')
if not raster_path.exists():
    print('Run `make run-pipeline` from the repository root first.')
else:
    with rasterio.open(raster_path) as src:
        velocity = src.read(1)
    plt.figure(figsize=(9, 5))
    plt.imshow(velocity, cmap='RdBu_r')
    plt.colorbar(label='Synthetic LOS velocity (mm/year)')
    plt.title('DEMO ONLY — synthetic planted hotspots')
    plt.show()
```

```python
ranked = Path('../outputs/ranked_assets.csv')
if ranked.exists():
    df = pd.read_csv(ranked)
    display(df.sort_values('risk_score', ascending=False).head(10))
else:
    print('Ranked output not found yet.')
```

## Interpretation

The synthetic field tests the software path. It cannot validate deformation in Greater Tunis. Production interpretation requires real HyP3/MintPy products, coherence filtering, reference-point documentation, sensitivity checks and independent ground validation.
